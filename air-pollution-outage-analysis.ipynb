## Abstract

This project investigates whether ambient air pollutant concentrations predict the severity of major power outages across the continental United States, and whether that relationship varies by climate region. While prior literature has established that outages worsen air quality (via diesel generator use and increased traffic), the inverse direction — pollution predicting outage severity — is substantially less studied. We merge two complementary public datasets: the CACES annual county-level air quality estimates (2000–2016) and the Purdue Major Power Outage dataset (1,534 events, 2000–2016), joined at the state × year level. Six EPA criteria pollutants (CO, NO₂, O₃, PM10, PM2.5, SO₂) serve as predictors, where log-transformed outage duration is the primary outcome.

Exploratory analysis reveals that all pollutants declined significantly over the study period. After partialing out year, modest positive partial correlations remain between pollutant levels and outage duration. These associations are stronger in state-years where severe weather dominates outage causes, supporting the hypothesis that a shared atmospheric driver simultaneously traps pollutants and stresses grid infrastructure. Regional similarities are very prominent, with some climate zones show consistent positive partial correlations across all pollutants, while others show near-zero or mixed results. KNN classification confirms that pollutant levels carry predictive signal for above/below-median outage severity in several regions.

Overall, the evidence supports a real but modest association between ambient pollution and outage severity, most plausibly explained by shared atmospheric confounding rather than a direct causal pathway from pollution to infrastructure failure. The findings motivate further research with finer spatial resolution and weather-controlling covariates.

## Background and Prior Work

&emsp;In the United States, air quality is monitored and defined by the Environmental Protection Agency (EPA) responsible for the protection of human health as well as environment. The EPA’s Air Quality Index (AQI) ranges from “Good” to “Hazardous” where the severity of pollutant diffusion across the atmosphere poses little risk to the public (“Good”) to everyone being increasingly likely to experience adverse effects from its exposure (“Hazardous”). The AQI’s index values intuitively rate air quality by having a higher AQI value meaning the greater the level of air pollution and vice versa.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) It is important to note however that the AQI is a composite measure and takes multiple air pollutants into account rather than just an individual pollutant concentration.

## Data

### Data overview

Our research question has two distinct components: (1) whether air pollutant levels predict power outage severity, and (2) whether these predictors vary by region. Answering it requires one dataset that measures ambient air pollution across U.S. states over time, and a separate dataset that records the severity, cause, and geographic location of power outage events over the same period. The CACES data has no outage information, and the Purdue data has no air quality measurements, so both are strictly necessary for this analysis. The analysis depends on merging them into a single state × year dataset where both pollution levels and outage metrics are jointly observed.

**CACES U.S. Air Quality Data** (https://www.caces.us/data) provides the air pollution side of the analysis. It supplies annual, county-level, population-weighted ambient concentration estimates forten pollutants — CO, NO₂, O₃, BC, PM10, PM2.5, SO₂, COA, HOA, and PNC — covering 1979–2020. Key columns are `fips` (county identifier), `pollutant`, `year`, `pred_wght` (the population-weighted concentration estimate), `state_abbr`, `lat`, and `lon`. Estimates are derived from EPA monitoring networks combined with land-use regression models, providing coverage even for counties without physical monitoring stations. Population-weighting means the estimate reflects the exposure experienced by an average resident of that county rather than a simple geographic average, making it more meaningful for health and policy analysis than a raw spatial mean.

Because the outage dataset records location only at the state level, we aggregate CACES county estimates to annual state means using `state_abbr` and `year` as keys. This loses within-state spatial variation but produces a common geographic identifier for merging. Four of the ten pollutants (BC, COA, HOA, PNC) have sparse coverage that does not overlap with the outage dataset's 2000–2016 window after merging, and are excluded from modeling — leaving CO, NO₂, O₃, PM10, PM2.5, and SO₂ as the six predictors used in analysis. These six are also the primary components of the EPA's AQI described in the background, making them the most policy-relevant measures available.

**Purdue Major Power Outage Risks in the U.S.** (https://engineering.purdue.edu/LASCI/research-data/outages) provides the outage severity side of the analysis. It records 1,534 major power outage events in the contiguous U.S. from January 2000 to July 2016, with one row per event. The variables we use as outage severity outcomes are `OUTAGE.DURATION` (minutes) and `CUSTOMERS.AFFECTED` — directly measuring how long and how widely each outage impacted the grid. `DEMAND.LOSS.MW` provides a third severity measure reflecting peak grid stress. `CAUSE.CATEGORY` is critical for our analysis because severe weather is the only cause with a plausible atmospheric connection to air quality — we use it to subset analyses and test whether the pollutant-outage association is stronger in weather-caused events specifically. `CLIMATE.REGION` (nine U.S. climate zones) is the primary variable for the regional heterogeneity component of our question, allowing us to test whether pollutant-outage associations differ across climate zones. Sector-level electricity prices (`RES.PRICE`, `COM.PRICE`, `IND.PRICE`, `TOTAL.PRICE`) serve as economic control variables rather than outcomes, since they reflect state energy market conditions rather than outage-specific severity.

Because outage events are recorded individually but CACES is annual and state-level, we aggregate outage events to state × year means and counts before merging. The two datasets are joined on `state_abbr` and `year` using an inner join, retaining only state-years present in restricting the analysis to 2000–2016 across the contiguous 48 states. The merged dataset is the sole input to all EDA and modeling: CACES contributes the six pollutant concentration columns, and the Purdue data contributes the outage severity metrics, cause breakdown, and climate region classification.

### Dataset #1 - CACES U.S. Air Quality

```python
caces_raw = pd.read_excel('data/00-raw/caces_us_data.xlsx')
caces = caces_raw.copy()
caces.columns = caces.columns.str.strip().str.lower()

# Restrict to the 6 well-covered pollutants
pollutants = ["co", "no2", "o3", "pm10", "pm25", "so2"]
caces_filtered = caces[caces['pollutant'].str.lower().isin(pollutants)].copy()
caces_filtered['pollutant'] = caces_filtered['pollutant'].str.lower()

# Aggregate to population-weighted mean across counties
state_yr_long = (
    caces_filtered
    .groupby(['year', 'state_abbr', 'pollutant'])['pred_wght']
    .mean()
    .reset_index()
)

state_yr = (
    state_yr_long
    .pivot_table(index = ['year','state_abbr'], columns = 'pollutant', values = 'pred_wght')
    .reset_index()
)

state_yr.columns.name = None
state_yr = state_yr.rename(columns = {p: f'poll_{p}' for p in pollutants})
state_yr['state_abbr'] = state_yr['state_abbr'].str.strip().str.upper()
state_yr['year'] = state_yr['year'].astype(int)

state_yr.to_excel('data/02-processed/caces_state_year.xlsx', index = False)
```

### Dataset #2  - Purdue Major Power Outages

```python
purdue_raw = pd.read_excel(
    'data/00-raw/power_outage_data.xlsx',
    sheet_name = 0, skiprows = 5, header = 0
)
purdue_raw.columns = purdue_raw.columns.str.strip()
outage = purdue_raw[pd.to_numeric(purdue_raw['YEAR'], errors = 'coerce').notna()].copy()

# Rename columns
col_map = {
    'YEAR'                    : 'year',
    'MONTH'                   : 'month',
    'U.S._STATE'              : 'state',
    'POSTAL.CODE'             : 'state_abbr',
    'NERC.REGION'             : 'nerc_region',
    'CLIMATE.REGION'          : 'climate_region',
    'CLIMATE.CATEGORY'        : 'climate_category',
    'ANOMALY.LEVEL'           : 'anomaly_level',
    'CAUSE.CATEGORY'          : 'cause_category',
    'CAUSE.CATEGORY.DETAIL'   : 'cause_detail',
    'OUTAGE.DURATION'         : 'outage_duration_min',
    'DEMAND.LOSS.MW'          : 'demand_loss_mw',
    'CUSTOMERS.AFFECTED'      : 'customers_affected',
    'RES.PRICE'               : 'price_residential',
    'COM.PRICE'               : 'price_commercial',
    'IND.PRICE'               : 'price_industrial',
    'TOTAL.PRICE'             : 'price_total',
    'TOTAL.SALES'             : 'total_sales_mwh',
    'TOTAL.CUSTOMERS'         : 'total_customers',
    'PC.REALGSP.STATE'        : 'real_gsp_usd',
    'PC.REALGSP.CHANGE'       : 'real_gsp_change',
    'POPULATION'              : 'population',
    'POPPCT_URBAN'            : 'pct_urban_pop',
    'POPDEN_URBAN'            : 'pop_density_urban',
    'OUTAGE.START.DATE'       : 'outage_start_date',
    'OUTAGE.RESTORATION.DATE' : 'outage_restore_date',
}
col_map = {k: v for k, v in col_map.items() if k in outage.columns}
outage = outage.rename(columns = col_map)

# Numeric coercion 
num_cols = [
    'year', 'month', 'outage_duration_min', 'customers_affected',
    'demand_loss_mw', 'price_residential', 'price_commercial',
    'price_industrial', 'price_total', 'total_sales_mwh',
    'total_customers', 'real_gsp_usd', 'population', 'pop_density_urban',
]
for c in num_cols:
    if c in outage.columns:
        outage[c] = pd.to_numeric(outage[c], errors = 'coerce')

outage = outage.dropna(subset = ['state', 'year'])
outage['year'] = outage['year'].astype(int)
outage['state_abbr'] = outage['state_abbr'].str.strip().str.upper()
outage['outage_duration_hr'] = outage['outage_duration_min'] / 60
outage[['year','state','state_abbr','cause_category',
        'outage_duration_min','price_total']].head()

outage.to_csv('data/01-interim/outage_cleaned.csv', index = False)

# Aggregate to state × year for merging 
outage_agg = (
    outage
    .groupby(['year', 'state_abbr'])
    .agg(
        n_outages           = ('state_abbr',          'count'),
        total_duration_min  = ('outage_duration_min',  'sum'),
        mean_duration_min   = ('outage_duration_min',  'mean'),
        median_duration_min = ('outage_duration_min',  'median'),
        total_customers_aff = ('customers_affected',   'sum'),
        mean_demand_loss_mw = ('demand_loss_mw',       'mean'),
        mean_price_total    = ('price_total',          'mean'),
        mean_price_res      = ('price_residential',    'mean'),
        mean_price_com      = ('price_commercial',     'mean'),
        mean_price_ind      = ('price_industrial',     'mean'),
        mean_real_gsp       = ('real_gsp_usd',         'mean'),
    )
    .reset_index()
)

# Join state full name and climate region back for labeling
state_meta = (outage[['state_abbr','state','climate_region','nerc_region']]
              .drop_duplicates('state_abbr')
              .dropna(subset = ['state_abbr']))
outage_agg = outage_agg.merge(state_meta, on = 'state_abbr', how = 'left')

outage_agg.to_excel('data/02-processed/outage_state_year.xlsx', index = False)

# Merge the two processed datasets on state × year
caces_state_year  = pd.read_excel('data/02-processed/caces_state_year.xlsx')
outage_state_year = pd.read_excel('data/02-processed/outage_state_year.xlsx')

for df in [caces_state_year, outage_state_year]:
    df['state_abbr'] = df['state_abbr'].str.strip().str.upper()
    df['year'] = df['year'].astype(int)

merged = pd.merge(outage_state_year, caces_state_year,
                  on = ['year', 'state_abbr'], how = 'inner')
merged['log_mean_duration'] = np.log1p(merged['mean_duration_min'])
merged.to_csv('data/02-processed/merged_outage_air_quality.csv', index = False)
```

### Section 1 — Air Pollutant Trends Over Time

Before examining any relationship with outages, we establish whether pollutant concentrations trend over time. A shared downward trend in both pollutants and any outage metric would produce spurious correlations rather than a meaningful association, so we'll instead quantify each pollutant's trend with a linear regression on year.

```python
# Figure 1: National average pollutant concentrations with OLS trend lines
national = merged.groupby('year')[poll_cols].mean().reset_index()
colors = sns.color_palette('tab10', len(poll_cols))

fig, axes = plt.subplots(2, 3, figsize = (14, 7), sharex = True)

for ax, poll, color in zip(axes.flat, poll_cols, colors):
    y = national[poll]
    ax.plot(national['year'], y, marker = 'o', lw = 2, color = color, markersize = 4)
    ax.fill_between(national['year'], y, alpha = 0.1, color = color)

    slope, intercept, r, p, _ = stats.linregress(national['year'], y)
    xs = np.array([national['year'].min(), national['year'].max()])
    ax.plot(xs, slope * xs + intercept, '--', color = 'gray', lw = 1.5)

    label = poll_labels.get(poll, poll)
    unit = label.split('(')[-1].replace(')', '') if '(' in label else ''
    ax.set_title(label.split(' ')[0])
    ax.set_ylabel(unit)
    ax.tick_params(axis = 'x', rotation = 40)
    ax.annotate(f'r\u00b2 = {r**2:.2f}, p = {p:.3f}',
                xy = (0.05, 0.08), xycoords = 'axes fraction',
                fontsize = 8, color = 'gray')

fig.suptitle('Figure 1: National Average Air Pollutant Concentrations, 2000\u20132016\n'
             '(dashed = OLS trend; r\u00b2 and p annotated per panel)', y = 1.01)
plt.tight_layout()
plt.show()
```

Every pollutant declined significantly over 2000–2016 (all p < 0.01), reflecting Clean Air Act enforcement, coal plant retirements, and vehicle emissions improvements. SO₂ and CO fell the most steeply in proportional terms; O₃ is the most volatile year-to-year because it is a secondary pollutant formed atmospherically rather than emitted directly.

This downward trend is the central concern for our analysis. Any outage metric that also moves over time will appear correlated with pollutants for spurious reasons, so we'll address this by computing partial correlations controlling for year in Section 3, and by including year as a covariate in all predictive models.

### Section 2 — Outage Severity: Distribution and Cause Breakdown

### Section 4 — Regional Heterogeneity

The second part of our research question asks whether the pollutant–outage association varies by climate region. We use two approaches: within-region partial correlations and KNN classification accuracy by region.

```python
# Figure 6: Within-region partial correlation heatmap
regions = merged['climate_region'].dropna().unique()
results = []

for poll in poll_cols:
    for region in regions:
        sub = merged[merged['climate_region'] == region][
            [poll, 'log_mean_duration', 'year']].dropna()
        if len(sub) >= 5:
            pr, pp = partial_r_year(sub, poll, 'log_mean_duration')
            results.append({
                'pollutant': poll.replace('poll_', '').upper(),
                'region': region,
                'partial_r': round(pr, 2),
                'p': round(pp, 3),
                'n': len(sub)
            })

corr_regional = pd.DataFrame(results)
corr_pivot = corr_regional.pivot(index = 'region', columns = 'pollutant',
                                  values = 'partial_r')

fig, ax = plt.subplots(figsize = (12, 6))
sns.heatmap(corr_pivot, annot = True, fmt = '.2f', center = 0,
            vmin = -0.6, vmax = 0.6,
            cmap = 'coolwarm', linewidths = 0.4, ax = ax, annot_kws = {'size': 10})
ax.set_title('Figure 6: Partial r (controlling for year): Pollutants vs. Log Outage Duration\n'
             'by Climate Region  —  positive = higher pollution -> longer outages',
             pad = 12)
ax.set_xlabel('Pollutant')
ax.set_ylabel('Climate Region')
plt.tight_layout()
plt.show()
```

The within-region correlations vary substantially across climate zones, directly supporting the regional heterogeneity component of our hypothesis. Regions with near-zero or negative correlations suggest that local outage causes are driven by factors largely independent of ambient air quality.

```python
# Figure 7: KNN classification accuracy by climate region
median_dur = merged['log_mean_duration'].median()
merged['high_severity'] = (merged['log_mean_duration'] > median_dur).astype(int)

feature_cols = poll_cols + ['year']
cv = StratifiedKFold(n_splits = 5, shuffle = True, random_state = 39)

# Select best k on full dataset
clf_all = merged[feature_cols + ['high_severity']].dropna()
X_all = StandardScaler().fit_transform(clf_all[feature_cols])
y_all = clf_all['high_severity'].values

k_scores = [cross_val_score(KNeighborsClassifier(n_neighbors = k),
                             X_all, y_all, cv = cv, scoring = 'accuracy').mean()
            for k in range(1, 16)]
best_k = np.argmax(k_scores) + 1
overall_acc = max(k_scores)

# Per-region accuracy
region_accs = {}
clf_data = merged[feature_cols + ['high_severity', 'climate_region']].dropna()

for region in clf_data['climate_region'].unique():
    sub = clf_data[clf_data['climate_region'] == region]
    if len(sub) < 15:
        continue
    Xr  = StandardScaler().fit_transform(sub[feature_cols])
    yr  = sub['high_severity'].values
    k_r = min(best_k, len(sub) // 5)
    acc = cross_val_score(
        KNeighborsClassifier(n_neighbors = k_r), Xr, yr,
        cv = StratifiedKFold(n_splits = 5, shuffle = True, random_state = 39),
        scoring = 'accuracy').mean()
    region_accs[region] = {'accuracy': acc, 'n': len(sub)}

region_acc_df = pd.DataFrame(region_accs).T.sort_values('accuracy', ascending=False)

fig, ax = plt.subplots(figsize = (11, 4))
bar_colors = ['tomato' if v > overall_acc else 'steelblue'
              for v in region_acc_df['accuracy']]
ax.bar(region_acc_df.index, region_acc_df['accuracy'],
       color = bar_colors, edgecolor = 'white')
ax.axhline(0.5, color = 'gray', ls = ':', lw = 1.5, label = 'Chance (0.50)')
ax.axhline(overall_acc, color = 'black', ls = '--', lw = 1.5,
           label = f'Overall CV accuracy ({overall_acc:.3f})')
ax.set_ylabel('5-Fold CV Accuracy')
ax.set_title('Figure 7: KNN Classification Accuracy by Climate Region\n'
             '(predicting above/below median outage duration from pollutant levels + year;\n'
             'red = above overall average, blue = below)', pad = 10)
ax.tick_params(axis = 'x', rotation = 40)
ax.legend(fontsize = 9)
plt.tight_layout()
plt.show()
```

KNN classification accuracy varies by region, confirming that pollutant levels are meaningfully more predictive of outage severity in some climate zones than others. This difference is detectable with a nonparametric classifier, not just correlation coefficients, providing convergent evidence for regional heterogeneity.

### Analysis 1 — Multiple OLS Regression

The EDA established that pollutants have modest positive partial correlations with log outage duration after controlling for year. Regression formalizes this, where we fit a single model with all six pollutants and year as predictors, and examine which coefficients are statistically significant. We run the model twice - once on all state-years, and once restricted to weather-dominated state-years (>50% of outages caused by severe weather) - to test whether the signal is stronger in the subset where an atmospheric mechanism is most plausible.
 
Predictors: poll_co, poll_no2, poll_o3, poll_pm10, poll_pm25, poll_so2, year  
Outcome: log(mean outage duration + 1)  
Method: OLS with standardized predictors so coefficients are comparable across pollutants

```python
merged = pd.read_csv('data/02-processed/merged_outage_air_quality.csv')
outage = pd.read_csv('data/01-interim/outage_cleaned.csv')

merged['log_mean_duration'] = np.log1p(merged['mean_duration_min'])
outage['weather_caused'] = (outage['cause_category'] == 'severe weather')
 
poll_cols = [c for c in merged.columns if c.startswith('poll_')]
 
# Build the weather-fraction column
weather_frac = (
    outage[outage['weather_caused']]
    .groupby(['year', 'state_abbr'])
    .size()
    .reset_index(name = 'n_weather')
)
total_out = (
    outage.groupby(['year', 'state_abbr'])
    .size()
    .reset_index(name = 'n_total')
)
weather_info = weather_frac.merge(total_out, on = ['year', 'state_abbr'])
weather_info['pct_weather'] = weather_info['n_weather'] / weather_info['n_total']
 
merged_w = merged.merge(
    weather_info[['year', 'state_abbr', 'pct_weather']],
    on = ['year', 'state_abbr'], how = 'left'
)
merged_w['pct_weather'] = merged_w['pct_weather'].fillna(0)
 
# Helper: run OLS and return a summary DataFrame 
def run_ols(df, feature_cols, outcome_col, label):
    """
    Fit OLS on standardized predictors.
    Returns a DataFrame of coefficients, standard errors, t-stats, and p-values,
    plus the model R².
    """
    sub   = df[feature_cols + [outcome_col]].dropna()
    X_raw = sub[feature_cols].values
    y     = sub[outcome_col].values
 
    # Standardize so coefficients are on the same scale (beta coefficients)
    scaler = StandardScaler()
    X = scaler.fit_transform(X_raw)
 
    # Fit with sklearn for predictions; compute stats manually
    model = LinearRegression().fit(X, y)
    y_hat = model.predict(X)
    r2    = r2_score(y, y_hat)
 
    n, p = X.shape
    residuals = y - y_hat
    sigma2    = residuals.var(ddof = p + 1)          # MSE
    XtX_inv   = np.linalg.pinv(X.T @ X)
    se        = np.sqrt(np.diag(XtX_inv) * sigma2)   # standard errors
    t_stats   = model.coef_ / se
    # Two-tailed p-values from t-distribution
    p_values  = [2 * (1 - stats.t.cdf(abs(t), df = n - p - 1)) for t in t_stats]
 
    results = pd.DataFrame({
        'predictor':  feature_cols,
        'beta_coef':  model.coef_.round(4),
        'std_error':  se.round(4),
        't_stat':     t_stats.round(3),
        'p_value':    [round(p, 4) for p in p_values],
        'significant': ['***' if p < 0.001 else '**' if p < 0.01
                        else '*' if p < 0.05 else '' for p in p_values]
    })
    print(f"\n{'='*55}")
    print(f"OLS Results — {label}  (n = {n}, R² = {r2:.3f})")
    print(f"{'='*55}")
    print(results.to_string(index = False))
    return results, r2
 
feature_cols = poll_cols + ['year']
 
results_all, r2_all = run_ols(
    merged_w, feature_cols, 'log_mean_duration', 'All state-years')
 
results_wx, r2_wx = run_ols(
    merged_w[merged_w['pct_weather'] > 0.5],
    feature_cols, 'log_mean_duration', 'Weather-dominated state-years')
```

```python
# Figure 10: coefficient plot comparing both models side by side
poll_labels_short = {
    'poll_co':   'CO',   'poll_no2': 'NO\u2082',
    'poll_o3':   'O\u2083',  'poll_pm10': 'PM10',
    'poll_pm25': 'PM2.5', 'poll_so2': 'SO\u2082',
    'year':      'Year'
}
 
fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)
datasets  = [
    (results_all, r2_all,  'All State-Years'),
    (results_wx,  r2_wx,   'Weather-Dominated State-Years'),
]
 
for ax, (res, r2, title) in zip(axes, datasets):
    poll_only = res[res['predictor'].isin(poll_cols)].copy()
    poll_only['label'] = poll_only['predictor'].map(poll_labels_short)
    colors = ['tomato' if b > 0 else 'steelblue' for b in poll_only['beta_coef']]
 
    bars = ax.barh(poll_only['label'], poll_only['beta_coef'],
                   color = colors, edgecolor = 'white')
    ax.axvline(0, color = 'gray', lw=0.8)
 
    # Add significance stars
    for bar, sig in zip(bars, poll_only['significant']):
        if sig:
            x = bar.get_width()
            offset = 0.002 if x >= 0 else -0.002
            ha = 'left' if x >= 0 else 'right'
            ax.text(x + offset, bar.get_y() + bar.get_height() / 2,
                    sig, va = 'center', ha = ha, fontsize = 11)
 
    ax.set_xlabel('Standardized Beta Coefficient')
    ax.set_title(f'{title}\n(R\u00b2 = {r2:.3f};  * p<0.05, ** p<0.01, *** p<0.001)')
 
axes[0].set_ylabel('Pollutant')
fig.suptitle(
    'Figure 10: OLS Regression — Standardized Pollutant Coefficients\n'
    'Predicting Log Mean Outage Duration (controlling for year)',
    y = 1.02
)
plt.tight_layout()
plt.show()
```

The OLS model on all state-years explains a moderate share of variance in log outage duration (R² = 0.177). After controlling for year, PM2.5 emerges as the strongest and most significant positive predictor (β = 0.606, p < 0.001), followed by PM10 (β = 0.247, p < 0.05). O₃ is the only pollutant with a significant *negative* coefficient (β = −0.328, p < 0.01), which is consistent with O₃ being a secondary pollutant formed atmospherically, as its concentration reflects sunlight and temperature patterns rather than direct emissions and may behave differently from combustion-related pollutants in this context. CO, NO₂, and SO₂ do not reach significance after controlling for the other pollutants and year.

The weather-dominated subset tells a different story (R² = 0.055). No individual pollutant reaches significance, and the PM2.5 coefficient actually flips negative (β = −0.148), reversing its direction from the full model. The overall model fit drops substantially. This is surprising given the EDA finding that pollutant–duration associations were stronger in weather-dominated state-years. The regression result suggests that once all pollutants are included simultaneously and collinearity is accounted for, the signal in the weather subset is not robust. The smaller sample size (n = 203 vs. 389) also reduces statistical power, making it harder to detect effects that may genuinely exist.

Taken together, the OLS results suggest that particulate matter — particularly PM2.5 — is the pollutant most consistently associated with outage severity in the full dataset, but that this association does not cleanly survive subsetting to weather-dominated state-years when all predictors are considered jointly.

### Analysis 2 — Permutation Test: Are Pollutant–Outage Correlations Stronger Than Chance?
 
The OLS regression shows which pollutants have significant coefficients, but significance depends on distributional assumptions (normality of residuals, etc.) that may not hold given the skewed outage data. A permutation test makes no such assumptions: it asks "how often would we see a correlation this large if the pairing between pollution levels and outage duration were random?"
 
Procedure:
1. Compute the observed partial correlation (controlling for year) between each pollutant and log outage duration.
2. Randomly shuffle the pollutant column 5,000 times, recomputing the partial correlation each time to build a null distribution.
3. The empirical p-value is the fraction of permuted correlations that are as extreme as or more extreme than the observed value.
 
We run this for all six pollutants and display the null distributions alongside the observed values. This validates (or challenges) the OLS significance results using a non-parametric approach.

```python
n_perm = 5000
rng    = np.random.default_rng(39)
 
obs_r  = {}
null_r = {}
emp_p  = {}
 
analysis_df = merged[poll_cols + ['log_mean_duration', 'year']].dropna().copy()
 
for poll in poll_cols:
    sub = analysis_df[[poll, 'log_mean_duration', 'year']].dropna()
 
    r_obs, _    = partial_r_year(sub, poll, 'log_mean_duration')
    obs_r[poll] = r_obs
 
    # Null distribution: shuffle the pollutant column, recompute partial r
    null_dist = []
    poll_vals = sub[poll].values.copy()
    for _ in range(n_perm):
        shuffled       = sub.copy()
        shuffled[poll] = rng.permutation(poll_vals)
        r_perm, _      = partial_r_year(shuffled, poll, 'log_mean_duration')
        null_dist.append(r_perm)
 
    null_r[poll] = np.array(null_dist)
 
    emp_p[poll] = np.mean(np.abs(null_dist) >= abs(r_obs))
 
print(f"\nPermutation Test Results (n_permutations={n_perm})\n{'='*50}")
print(f"{'Pollutant':<12} {'Observed r':>12} {'Emp. p-value':>14} {'Sig':>6}")
print('-' * 50)
for poll in poll_cols:
    name = poll.replace('poll_', '').upper()
    sig  = ('***' if emp_p[poll] < 0.001 else
            '**'  if emp_p[poll] < 0.01  else
            '*'   if emp_p[poll] < 0.05  else '')
    print(f"{name:<12} {obs_r[poll]:>12.4f} {emp_p[poll]:>14.4f} {sig:>6}")
```

```python
# Figure 11: null distributions with observed r marked
poll_labels_full = {
    'poll_co':   'CO',   'poll_no2': 'NO\u2082',
    'poll_o3':   'O\u2083',  'poll_pm10': 'PM10',
    'poll_pm25': 'PM2.5', 'poll_so2': 'SO\u2082',
}
 
fig, axes = plt.subplots(2, 3, figsize=(14, 7))
 
for ax, poll in zip(axes.flat, poll_cols):
    null = null_r[poll]
    r    = obs_r[poll]
    p    = emp_p[poll]
    name = poll_labels_full[poll]
 
    ax.hist(null, bins = 60, color = 'steelblue', edgecolor = 'white',
            lw = 0.3, alpha = 0.75, label = 'Null distribution')
    ax.axvline(r, color = 'tomato', lw = 2.5,
               label = f'Observed r = {r:.3f}')
    ax.axvline(-abs(r), color = 'tomato', lw = 2.5, ls = '--', alpha = 0.5)
 
    sig = ('***' if p < 0.001 else '**' if p < 0.01
           else '*' if p < 0.05 else 'n.s.')
    ax.set_title(f'{name}  (emp. p = {p:.3f}  {sig})')
    ax.set_xlabel('Permuted Partial r')
    ax.set_ylabel('Frequency')
    ax.legend(fontsize = 7)
 
fig.suptitle(
    f'Figure 11: Permutation Test Null Distributions vs. Observed Partial r\n'
    f'(controlling for year;  n = {n_perm:,} permutations;\n'
    f'dashed red = mirrored observed value for two-tailed reference)',
    y = 1.02
)
plt.tight_layout()
plt.show()
```

The permutation test provides non-parametric validation of the correlation results without assuming normally distributed residuals. Of the six pollutants, PM2.5 and PM10 are the only ones whose observed partial correlations fall reliably in the tail of the null distribution: PM2.5 shows an observed partial r of 0.261 (emp. p < 0.001) and PM10 shows r = 0.158 (emp. p = 0.002). Both are significant at the 0.01 level, confirming that their associations with outage duration are stronger than would be expected by random pairing of pollution levels and outage outcomes.

CO, NO₂, O₃, and SO₂ do not reach significance under the permutation test (all emp. p > 0.12). Their observed partial r values fall well within the bulk of the null distribution, indicating that their apparent associations are consistent with chance variation after controlling for year.

This partially aligns with the OLS results — both methods identify PM2.5 and PM10 as the strongest predictors — but the permutation test does not corroborate OLS's finding of a significant negative O₃ coefficient. That discrepancy likely reflects the difference between what each method is measuring: OLS estimates the conditional effect of O₃ holding all other pollutants constant (where multicollinearity among pollutants can inflate or distort individual coefficients), while the permutation test evaluates each pollutant's bivariate partial correlation with the outcome independently of the others. The fact that both PM2.5 and PM10 survive both tests strengthens confidence that their associations are real rather than artifacts of model specification.

Overall, the permutation test converges on a consistent conclusion: among the six EPA criteria pollutants examined, particulate matter — specifically PM2.5 and PM10 — shows the most reliable statistical association with power outage severity, robust to both parametric and non-parametric testing approaches.

## Discussion and Conclusion

This project investigated whether ambient air pollutant concentrations predict the severity of major power outages across the continental United States, and whether that association varies by climate region. Using a merged state × year dataset combining CACES air quality estimates and Purdue power outage records for 2000–2016, we found modest but consistent positive associations between pollutant levels and log-transformed outage duration, most clearly for PM2.5 and PM10. These associations survived partial correlation controls for year, though they were substantially attenuated compared to raw correlations, confirming that a shared downward time trend partially confounds the raw signal.

Our background established a well-documented causal pathway running from outages to worsening air quality (via diesel generator use). This project tested the inverse: whether ambient pollution levels predict subsequent outage severity. The modest positive associations we found are consistent with the hypothesis that shared atmospheric drivers — particularly stagnant high-pressure systems that simultaneously trap pollutants and create heat stress on the grid — produce a detectable correlation even in the absence of a direct causal mechanism from pollution to infrastructure failure. The amplification of associations in weather-dominated state-years (Figure 8) is the strongest evidence for this atmospheric confounding interpretation.

The within-region correlation heatmap (Figure 6) and KNN classification results (Figure 7) together demonstrate that the pollutant–outage association is not uniform across the U.S. This supports the regional component of our hypothesis and suggests that the atmospheric confounding mechanism operates more reliably in certain climate zones than others. Regions where pollution and severe weather stress tend to co-occur — particularly those with hot, humid summers — show the clearest signal.

Several important limitations temper these findings. First, state-level aggregation masks within-state spatial variation in both pollution exposure and grid reliability; a county-level analysis might reveal sharper associations or important heterogeneity currently hidden by averaging. Second, the Purdue dataset records only *major* outages, potentially underrepresenting outage severity patterns in rural or less-populated areas. Third, while we control for year as a linear trend, other time-varying confounders (economic conditions, grid investment patterns) are not fully accounted for. Fourth, the analysis cannot establish causation: the shared atmospheric driver hypothesis is plausible but not identified from observational data alone.

Future work could incorporate weather station data (temperature, humidity, wind) as explicit covariates to directly test the atmospheric confounding mechanism rather than relying on weather-cause categorization as a proxy. A county-level analysis joining CACES data to a more granular outage dataset would also substantially improve spatial precision. Longitudinal panel models with state fixed effects would better isolate within-state variation from cross-sectional confounding.
